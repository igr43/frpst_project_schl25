# VL-OFF — Подготовка данных (case)
В этом ноутбуке показан процесс предварительной подготовки данных по отключениям коммунальных ресурсов во Владивостоке.

**Содержимое:**
- Обзор таблиц в БД
- Загрузка таблиц
- Преобразование дат и расчёт длительности
- Разбор координат зданий
- Подсчёт затронутых зданий и объединение данных
- Сохранение очищённого CSV


```
# Подключение и обзор таблиц
import sqlite3, pandas as pd
conn = sqlite3.connect(r"/mnt/data/Кейс_Аналитика.db")
tables = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table' ORDER BY name;", conn)
print(tables)
```

```
# Загрузка таблиц в DataFrame
dfs = {}
for tbl in tables['name']:
    dfs[tbl] = pd.read_sql(f"SELECT * FROM "{tbl}";", conn)
    print(tbl, dfs[tbl].shape)
```

```
# Преобразование дат в таблице blackouts и расчет длительности
blackouts = dfs['blackouts'].copy()
blackouts['start_date'] = pd.to_datetime(blackouts['start_date'], errors='coerce', infer_datetime_format=True)
blackouts['end_date'] = pd.to_datetime(blackouts['end_date'], errors='coerce', infer_datetime_format=True)
blackouts['duration_hours'] = (blackouts['end_date'] - blackouts['start_date']).dt.total_seconds()/3600
blackouts['duration_days'] = blackouts['duration_hours']/24
blackouts.head()
```

```
# Подсчет числа затронутых зданий (если есть таблица blackouts_buildings)
if 'blackouts_buildings' in dfs:
    bb = dfs['blackouts_buildings']
    cnt = bb.groupby('blackout_id')['building_id'].nunique().reset_index().rename(columns={'building_id':'affected_buildings_count'})
    blackouts = blackouts.merge(cnt, how='left', left_on='id', right_on='blackout_id')
    blackouts['affected_buildings_count'] = blackouts['affected_buildings_count'].fillna(0).astype(int)
    print('Добавлен affected_buildings_count')
```

```
# Попытка разбора координат в таблице buildings
if 'buildings' in dfs and 'coordinates' in dfs['buildings'].columns:
    b = dfs['buildings']
    import re, numpy as np
    def parse_coords(x):
        if pd.isna(x): return (np.nan, np.nan)
        s = str(x).strip()
        m = re.search(r'POINT\(\s*([\-\d\.]+)\s+([\-\d\.]+)\s*\)', s)
        if m: return (float(m.group(2)), float(m.group(1)))
        m = re.search(r'^\s*([\-\d\.]+)\s*[, ]\s*([\-\d\.]+)\s*$', s)
        if m: return (float(m.group(1)), float(m.group(2)))
        return (np.nan, np.nan)
    coords = b['coordinates'].apply(parse_coords)
    b[['lat','lon']] = pd.DataFrame(coords.tolist(), index=b.index)
    b[['coordinates','lat','lon']].head()
```

```
# Сохранение очищённого CSV
blackouts.to_csv(r"/mnt/data/cleaned_blackouts.csv", index=False)
print('Saved to /mnt/data/cleaned_blackouts.csv')
```
